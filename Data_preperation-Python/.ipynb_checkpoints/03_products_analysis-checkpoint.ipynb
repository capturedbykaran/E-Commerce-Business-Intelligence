**DATA_CLEANING**

```python
products["product_category_name"] = products["product_category_name"].fillna("Unknown")
```

```python
products["product_name_lenght"] = products["product_name_lenght"].fillna(products["product_name_lenght"].median())
```

```python
products["product_description_lenght"] = products["product_description_lenght"].fillna(products["product_description_lenght"].median())
```

```python
products["product_photos_qty"] = products["product_photos_qty"].fillna(products["product_photos_qty"].median())
```

```python
products["product_weight_gm"] = products["product_weight_gm"].fillna(products["product_weight_gm"].median())
```

```python
# dimensions
cols = [
    "product_length_cm",
    "product_height_cm",
    "product_width_cm"
]

for col in cols:
    products[col] = products[col].fillna(products[col].median())
```

```python
products[products["product_weight_g"] == 0]
```

```python
products["product_weight_g"] = products["product_weight_g"].replace(0, np.nan)
```

```python
products.head()
```

```python
products.rename(columns={
    "product_name_lenght":"product_name_length",
    "product_description_lenght":"product_description_length",
    "product_weight_g" : "product_weight_gm"
}, inplace=True)
```

**Feature_Engineering**

```python
# VOLUME OF PRODUCTS
products["product_volume_cm3"] = (
    products["product_length_cm"] *
    products["product_width_cm"] *
    products["product_height_cm"]
)
```

```python
# WEIGHT CATEGORY
def weight_category(x):

    if x < 500:
        return "Light"

    elif x < 2000:
        return "Medium"

    else:
        return "Heavy"

products["weight_category"] = products["product_weight_gm"].apply(weight_category)
```

```python
# SIZE CATEGORY
def size(volume):

    if volume < 2000:
        return "Small"

    elif volume < 10000:
        return "Medium"

    else:
        return "Large"

products["size_category"] = products["product_volume_cm3"].apply(size)
```

```python
# PHOTO QUALITY
def photos(x):

    if x == 1:
        return "Low"

    elif x <= 3:
        return "Average"

    else:
        return "High"

products["photo_quality"] = (
    products["product_photos_qty"]
    .apply(photos)
)
```

```python
# SHIPPING TYPE
def shipping(row):

    if row["product_weight_gm"] < 500 and row["product_volume_cm3"] < 3000:
        return "Economy"

    elif row["product_weight_gm"] < 2000:
        return "Standard"

    else:
        return "Heavy Cargo"

products["shipping_class"] = products.apply(shipping, axis=1)
```
