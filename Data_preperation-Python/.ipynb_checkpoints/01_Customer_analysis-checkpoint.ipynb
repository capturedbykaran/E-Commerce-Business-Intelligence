```python
south = ["PR","SC","RS"]

southeast = ["SP","RJ","MG","ES"]

north = ["AC","AM","AP","PA","RO","RR","TO"]

northeast = ["AL","BA","CE","MA","PB","PE","PI","RN","SE"]

central_west = ["DF","GO","MT","MS"]

def region(state):
    if state in south:
        return "South"

    elif state in southeast:
        return "Southeast"

    elif state in north:
        return "North"

    elif state in northeast:
        return "Northeast"

    else:
        return "Central-West"

customers["region"] = customers["customer_state"].apply(region)
```
