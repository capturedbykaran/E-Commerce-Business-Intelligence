**FEATURE ENGINEERING**

```python
# total cost
order_items["total_item_cost"] = (
    order_items["price"]
    + order_items["freight_value"]
)
```

```python
# Shipping Percentage
order_items["shipping_percent"] = (
    order_items["freight_value"]
    / order_items["price"]
) * 100
```

```python
# Shipping Category
order_items["shipping_category"] = (
    order_items["freight_value"]
    .apply(
        lambda x:
        "Free" if x == 0
        else "Low" if x < 20
        else "Medium" if x < 50
        else "High"
    )
)
```

```python
# Price Category
order_items["price_category"] = (
    order_items["price"]
    .apply(
        lambda x:
        "Budget" if x < 50
        else "Standard" if x < 150
        else "Premium" if x < 500
        else "Luxury"
    )
)
```
